# Sinusoidal signals

in this notebook we are working on plotting sine wave

# mathematical formula:
$$
y = sine (2*pi*t/T)
$$
where:
- t is time 
- T is the period at which the signal repeats itself

```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.arange(0.0, 1, 0.001)
# t is a numpy array that contains values from 0 to 1 with a step of 0.001 and represents time in seconds.
s = np.sin(2*np.pi*t) 
# s is a numpy array that contains the sine of 2πt, which represents the amplitude of the sine wave at each time point.
# plot the sine wave
plt.plot(t, s)
plt.title('Sine Wave')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid()
# save the plot as a PNG file
plt.savefig('sine_wave.png') 

plt.show() 
```

# Add cosine wave

```python
c = np.cos(2*np.pi*t)
# c is a numpy array that contains the cosine of 2πt, which represents the amplitude of the cosine wave at each time point.
plt.plot(t, s, t, c)
# plot both sine and cosine waves on the same graph
plt.title('Sine and Cosine Waves')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
# add a legend to the plot
plt.legend(['Sine', 'Cosine'])
plt.grid()
# save the plot as a PDF file
# pdf is making conflicts
#plt.savefig('sine_cosine_waves.pdf')
```

# Sine waves at diffrent frequencies  

```python
def sinusoid(frequency):
  #  t = np.arange(0.0, duration, 1/sample_rate)
    s = np.sin(2 * np.pi * frequency * t)
    return s

# define the time array t globally to avoid redefining it in the sinusoid function

first_frequency = 1  # Frequency in Hz
second_frequency = 2  # Frequency in Hz
# define two frquencies for the sine waves
# duration = 1.0  # Duration in seconds   
# sample_rate = 1000  # Sample rate in Hz
# no need to define duration and sample_rate since t is already defined globally


s1 = sinusoid(first_frequency)
s2 = sinusoid(second_frequency)
# define two sine waves with different frequencies using the sinusoid function

plt.plot(t, s1, t, s2)
plt.title('Sine Waves at Different Frequencies')    
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend([f'Sine {first_frequency} Hz', f'Sine {second_frequency} Hz'])
plt.grid()
plt.savefig('sine_waves_different_frequencies.png')
```
