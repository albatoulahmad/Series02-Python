# Sinusoidal signals

in this notebook we are working on plotting sine wave

# mathematical formula:
$$
y = sine (2*pi*t/T)
$$
where:
- t is time 
- T is the period at which the signal repeats itself

```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.arange(0.0, 1, 0.001)
# t is a numpy array that contains values from 0 to 1 with a step of 0.001 and represents time in seconds.
s = np.sin(2*np.pi*t) 
# s is a numpy array that contains the sine of 2πt, which represents the amplitude of the sine wave at each time point.
# plot the sine wave
plt.plot(t, s)
plt.title('Sine Wave')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid()
# save the plot as a PNG file
plt.savefig('sine_wave.png') 

plt.show() 
```
